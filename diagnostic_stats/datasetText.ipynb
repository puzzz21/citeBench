```python
import json
citedPaperIds = []
for file in ["Biology","Chemistry","Computer_Science","Materials_Science","Medicine","Physics", "Mathematics"]:
    with open('lengthDataset/length_'+ file +'.json') as f:
        records = json.load(f)
        print(len(records))
        for record in records:
            print(len(record))
            for data in record:                    
                citedPaperIds.append(data["citedPaper"])
```
